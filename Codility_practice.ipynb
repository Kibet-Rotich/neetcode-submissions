```python
def solution(N: int):
  binary = str(bin(N)[2:])
  gap,maxgap = 0,0

  for i in range(len(binary)):
    if binary[i] == '0':
      gap+=1
    elif binary[i] == '1':
      maxgap = max(maxgap,gap)
      gap = 0

  return maxgap


print(solution(1042))
```

```python
def solution(A,K):
  size = len(A)
  if not A:
    return A

  rem = K%size

  return A[-rem:] + A[:-rem] if rem != 0 else  A

A = [3, 8, 9, 7, 6]
K = 3

print(solution(A,K))
```

```python
def solution(A):
  total = sum(A)
  suml,sumr = 0,total
  min_diff = float('inf')

  for i in range(len(A)-1):
    suml = suml+A[i]
    sumr = sumr-A[i]
    diff = abs(sumr-suml)
    if diff < min_diff:
      min_diff = diff
  return min_diff

A = [3,1,2,4,3]
print(solution(A))
print(solution([-1000, 1000]))
```

```python
def solution(X,A):
  myset = set()
  for i in range(len(A)):
    myset.add(A[i])
    if len(myset) == X:
      return i
  return -1

A = [1, 3, 1, 4, 2, 3, 5, 4]
X = 5
print(solution(X,A))
```

```python
def solution(N,A):
  counters = [0] *N
  currentmax = 0
  lastmaxreset = 0
  for x in A:

    if 1<=x<=N:
      index = x-1

      if counters[index]<lastmaxreset:
        counters[index] = lastmaxreset

      counters[index]+=1

      if counters[index]> currentmax:
        currentmax = counters[index]
    elif x == N+1:
      lastmaxreset = currentmax


  for i in range(N):
    if counters[i]< lastmaxreset:
      counters[i] = lastmaxreset

  return counters


N = 5
A = [3, 4, 4, 6, 1, 4, 4]
print(solution(N, A))

N = 1
A = [1, 2, 1]
print(solution(N, A))
```

```python
def solution(A):
  passingcars = 0
  eastcars = 0
  for i in range(len(A)):
    if A[i] == 0:
      eastcars += 1
    elif A[i] == 1:
      passingcars += eastcars
      if passingcars > 1000000000:
        return -1
  return passingcars


A = [0,1,0,1,1]
print(solution(A))
```

```python

def solution(A,B,K):
  counter = 0
  for i in range(A,B+1):
    if i%K ==0:
      counter += 1

  return counter

print(solution(6,11,2))
```
